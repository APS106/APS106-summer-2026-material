## Problem 1 — Pizza Party Splitter 🍕
**Concepts: Variables, arithmetic operators, `input()`, type conversion, string formatting**

Write a program that:
1. Asks the user for the **total pizza bill** (a decimal number).
2. Asks the user for the **number of people** splitting the bill (an integer).

## Problem 2 — Password Strength Checker 🔐
**Concepts: Booleans, `if/elif/else`, string methods (`.upper()`, `.lower()`, `.isdigit()`), `for` loop**

Write a function `check_password(password)` that rates a password as:
- `"Weak"` — fewer than 6 characters  
- `"Medium"` — at least 6 characters, but missing either an uppercase letter **or** a digit  
- `"Strong"` — at least 8 characters, contains at least one uppercase letter **and** at least one digit  

Then ask the user to enter a password and print its rating.

**Example:**
```python
check_password("hello")      # → "Weak"
check_password("helloworld") # → "Medium"  (no uppercase or digit)
check_password("Hello1world")# → "Strong"
```


````python
# ✅ Solution — Problem 2

def check_password(password):
    """
    (str) -> str
    Returns the strength rating of a password: 'Weak', 'Medium', or 'Strong'.
    """
    # Check for uppercase and digit by looping through characters
    has_upper = False
    has_digit = False
    
    for char in password:
        if char == char.upper() and char.isalpha():  # uppercase letter
            has_upper = True
        if char.isdigit():
            has_digit = True
    
    if len(password) < 6:
        return "Weak"
    elif len(password) >= 8 and has_upper and has_digit:
        return "Strong"
    else:
        return "Medium"


# Test the function
print(check_password("hello"))        # Weak
print(check_password("helloworld"))   # Medium
print(check_password("Hello1world"))  # Strong

# Interactive version
pw = input("Enter a password to check: ")
print("Password strength:", check_password(pw))
````

## Problem 3 — Collatz Sequence 🔢
**Concepts: Functions, `while` loop, conditionals, `return`, modulo `%`**

The **Collatz conjecture** says that for any positive integer `n`, repeatedly applying the following rule will eventually reach 1:
- If `n` is **even** → `n = n // 2`
- If `n` is **odd** → `n = n * 3 + 1`

Write a function `collatz_steps(n)` that returns **how many steps** it takes to reach 1.

Then write a second function `longest_collatz(limit)` that finds the starting number below `limit` that produces the **longest Collatz sequence**.

**Example:**
```python
collatz_steps(6)       # → 8  (6 → 3 → 10 → 5 → 16 → 8 → 4 → 2 → 1)
longest_collatz(20)    # → 18  (18 takes 20 steps, the most below 20)
```


````python
# ✅ Solution — Problem 3

def collatz_steps(n):
    """
    (int) -> int
    Returns the number of steps to reach 1 via the Collatz rule.
    """
    steps = 0
    while n != 1:
        if n % 2 == 0:      # even
            n = n // 2
        else:               # odd
            n = n * 3 + 1
        steps += 1
    return steps


def longest_collatz(limit):
    """
    (int) -> int
    Returns the starting number below limit with the longest Collatz sequence.
    """
    max_steps  = 0
    best_start = 1
    
    n = 1
    while n < limit:
        steps = collatz_steps(n)
        if steps > max_steps:
            max_steps  = steps
            best_start = n
        n += 1
    
    return best_start


# Tests
print(collatz_steps(6))        # 8
print(collatz_steps(27))       # 111
print(longest_collatz(20))     # 18
print(longest_collatz(100))    # 97
````

## Problem 4 — Text Analyzer 📊
**Concepts: `for` loop, string methods (`.lower()`, `.isalpha()`, `.isdigit()`), counting, functions**

Write a function `analyze_text(text)` that returns a formatted summary string containing:
- Total character count (including spaces)
- Number of **letters**
- Number of **digits**
- Number of **spaces**
- Number of **vowels** (a, e, i, o, u — case-insensitive)
- The **most common vowel** (if any)

**Example:**
```python
analyze_text("Hello World! 123")
```
```
Total characters : 16
Letters          : 10
Digits           : 3
Spaces           : 2
Vowels           : 3
Most common vowel: o
```


````python
# ✅ Solution — Problem 4

def analyze_text(text):
    """
    (str) -> str
    Returns a formatted analysis of the input text.
    """
    total    = len(text)
    letters  = 0
    digits   = 0
    spaces   = 0
    
    # Track count of each vowel
    vowels = 'aeiou'
    vowel_counts = {}
    for v in vowels:
        vowel_counts[v] = 0   # we haven't learned dicts yet, but this is a preview!
    
    # --- alternatively, without dicts, track them separately ---
    count_a = count_e = count_i = count_o = count_u = 0
    
    for char in text:
        if char.isalpha():
            letters += 1
        if char.isdigit():
            digits += 1
        if char == ' ':
            spaces += 1
        
        lower_char = char.lower()
        if lower_char == 'a': count_a += 1
        elif lower_char == 'e': count_e += 1
        elif lower_char == 'i': count_i += 1
        elif lower_char == 'o': count_o += 1
        elif lower_char == 'u': count_u += 1
    
    vowel_total = count_a + count_e + count_i + count_o + count_u
    
    # Find the most common vowel
    max_count = 0
    most_common = ''
    
    if count_a > max_count: max_count = count_a; most_common = 'a'
    if count_e > max_count: max_count = count_e; most_common = 'e'
    if count_i > max_count: max_count = count_i; most_common = 'i'
    if count_o > max_count: max_count = count_o; most_common = 'o'
    if count_u > max_count: max_count = count_u; most_common = 'u'
    
    if most_common == '':
        most_common = 'None'
    
    result = (
        "Total characters : " + str(total)       + "\n" +
        "Letters          : " + str(letters)     + "\n" +
        "Digits           : " + str(digits)      + "\n" +
        "Spaces           : " + str(spaces)      + "\n" +
        "Vowels           : " + str(vowel_total) + "\n" +
        "Most common vowel: " + most_common
    )
    return result


print(analyze_text("Hello World! 123"))
print()
print(analyze_text("APS106 is a great course!"))
````

## Problem 5 — Number Pyramid 🔺
**Concepts: Nested `for` loops, `range()`, string repetition, `print()`**

Write a function `print_pyramid(n)` that prints a **number pyramid** of height `n`.  
Each row `i` (1-indexed) should:
- Be **centred** using spaces
- Print the digits `1` up to `i` then back down to `1`

**Example — `print_pyramid(5)`:**
```
    1    
   121   
  12321  
 1234321 
123454321
```

**Hint:** Each row has `2*i - 1` number characters and `n - i` spaces on each side.


````python
# ✅ Solution — Problem 5

def print_pyramid(n):
    """
    (int) -> None
    Prints a centred number pyramid of height n.
    """
    for i in range(1, n + 1):
        # Left padding
        spaces = ' ' * (n - i)
        
        # Build the number part: 1 2 3 ... i ... 3 2 1
        row = ''
        for j in range(1, i + 1):       # ascending:  1 → i
            row += str(j)
        for j in range(i - 1, 0, -1):   # descending: i-1 → 1
            row += str(j)
        
        print(spaces + row + spaces)


print_pyramid(5)
print()
print_pyramid(7)
````

## Problem 6 — Number Guessing Game 🎲
**Concepts: `while` loop, `random` module, `input()`, `int()`, conditionals, keeping score**

Write a guessing game where:
1. The computer picks a **random integer between 1 and 50** (inclusive).
2. The user keeps guessing until they get it right.
3. After each wrong guess, print `"Too high!"` or `"Too low!"`.
4. When the user guesses correctly, print the number of attempts it took.
5. **Bonus:** If the user enters something that's not a valid integer, print `"Please enter a valid number."` and ask again (don't count it as an attempt).

**Example run:**
```
Guess a number between 1 and 50: 25
Too high!
Guess a number between 1 and 50: 12
Too low!
Guess a number between 1 and 50: 18
🎉 Correct! You got it in 3 attempts.
```


````python
# ✅ Solution — Problem 6

import random

def guessing_game():
    """
    (None) -> None
    Plays a number guessing game with feedback.
    """
    secret   = random.randint(1, 50)
    attempts = 0
    guessed  = False
    
    while not guessed:
        user_input = input("Guess a number between 1 and 50: ")
        
        # Bonus: validate input before counting attempt
        # Check if every character is a digit (handles positive integers)
        valid = True
        if len(user_input) == 0:
            valid = False
        else:
            for ch in user_input:
                if not ch.isdigit():
                    valid = False
        
        if not valid:
            print("Please enter a valid number.")
        else:
            guess    = int(user_input)
            attempts += 1
            
            if guess < secret:
                print("Too low!")
            elif guess > secret:
                print("Too high!")
            else:
                print("🎉 Correct! You got it in", attempts, "attempt(s).")
                guessed = True


guessing_game()
````

## Problem 7 — Run-Length Encoding 🗜️
**Concepts: `for` loop, string building, functions, string indexing**

**Run-length encoding (RLE)** compresses repeated characters:  
`"aaabbbccddddee"` → `"3a3b2c4d2e"`

Write two functions:
1. `rle_encode(s)` — compresses a string using run-length encoding.
2. `rle_decode(s)` — takes an encoded string and reconstructs the original.

**Rules for `rle_encode`:**
- If a character appears only once, output just the character (not `1a`, just `a`).

**Examples:**
```python
rle_encode("aaabbbccddddee")   # → "3a3b2c4d2e"
rle_encode("abcde")            # → "abcde"
rle_encode("aabbccaa")         # → "2a2b2ca")  ← wait: 2a2b2c2a
rle_decode("3a3b2c4d2e")       # → "aaabbbccddddee"
rle_decode("abcde")            # → "abcde"
```


````python
# ✅ Solution — Problem 7

def rle_encode(s):
    """
    (str) -> str
    Returns the run-length encoding of s.
    Single-character runs are not prefixed with 1.
    """
    if len(s) == 0:
        return ''
    
    encoded = ''
    count   = 1
    
    # Walk from index 1 onward, comparing to previous character
    i = 1
    while i < len(s):
        if s[i] == s[i - 1]:
            count += 1
        else:
            # Flush the previous run
            if count > 1:
                encoded += str(count) + s[i - 1]
            else:
                encoded += s[i - 1]
            count = 1
        i += 1
    
    # Flush the final run
    if count > 1:
        encoded += str(count) + s[-1]
    else:
        encoded += s[-1]
    
    return encoded


def rle_decode(s):
    """
    (str) -> str
    Decodes a run-length encoded string back to the original.
    """
    decoded = ''
    i = 0
    
    while i < len(s):
        # Collect a run of digit characters
        num_str = ''
        while i < len(s) and s[i].isdigit():
            num_str += s[i]
            i += 1
        
        # The next character is the letter
        if i < len(s):
            letter = s[i]
            i += 1
            if num_str == '':
                decoded += letter        # no count → appears once
            else:
                decoded += letter * int(num_str)
    
    return decoded


# Tests
print(rle_encode("aaabbbccddddee"))   # 3a3b2c4d2e
print(rle_encode("abcde"))            # abcde
print(rle_encode("aabbccaa"))         # 2a2b2c2a

print(rle_decode("3a3b2c4d2e"))       # aaabbbccddddee
print(rle_decode("abcde"))            # abcde
print(rle_decode("2a2b2c2a"))         # aabbccaa

# Round-trip check
original = "aaabbbccddddee"
print(rle_decode(rle_encode(original)) == original)  # True
````

## Problem 8 — Log File Analyzer 📂
**Concepts: File I/O (`open`, `with`, `.read()` / `for line in file`), string methods (`.strip()`, `.find()`, `.split()` — preview), loops, conditionals**

A server log file records events in this format:
```
INFO: Server started
WARNING: High memory usage
ERROR: Connection timeout
INFO: Request processed
ERROR: Disk full
INFO: Request processed
```

The code below **creates the log file** for you. After running it, write a function  
`analyze_log(filename)` that reads the file and prints:
- Total number of lines
- Count of `INFO`, `WARNING`, and `ERROR` entries
- All `ERROR` messages (just the message part, after the colon)

**Expected output:**
```
Total lines : 6
INFO        : 3
WARNING     : 1
ERROR       : 2

Error messages:
  - Connection timeout
  - Disk full
```


````python
# Run this cell first — it creates the log file
log_contents = """INFO: Server started
WARNING: High memory usage
ERROR: Connection timeout
INFO: Request processed
ERROR: Disk full
INFO: Request processed"""

with open("server.log", "w") as f:
    f.write(log_contents)

print("server.log created successfully.")
````

````python
# ✅ Solution — Problem 8

def analyze_log(filename):
    """
    (str) -> None
    Reads a log file and prints a summary of INFO, WARNING, and ERROR counts,
    plus the text of every ERROR message.
    """
    info_count    = 0
    warning_count = 0
    error_count   = 0
    total_lines   = 0
    error_messages = ''   # build up as a string (no lists yet!)
    
    with open(filename, 'r') as logfile:
        for line in logfile:
            line = line.strip()   # remove leading/trailing whitespace & newline
            
            if line == '':        # skip blank lines
                continue
            
            total_lines += 1
            
            # Identify the log level (everything before the colon)
            colon_idx = line.find(':')
            
            if colon_idx != -1:
                level   = line[:colon_idx]          # e.g. "INFO"
                message = line[colon_idx + 2:]      # e.g. "Server started"
            else:
                level   = line
                message = ''
            
            if level == 'INFO':
                info_count += 1
            elif level == 'WARNING':
                warning_count += 1
            elif level == 'ERROR':
                error_count += 1
                error_messages += '  - ' + message + '\n'
    
    # Print summary
    print("Total lines : " + str(total_lines))
    print("INFO        : " + str(info_count))
    print("WARNING     : " + str(warning_count))
    print("ERROR       : " + str(error_count))
    
    if error_count > 0:
        print("\nError messages:")
        print(error_messages, end='')


analyze_log("server.log")
````

## 🌟 Bonus — Combining Everything: CSV Grade Processor
**Concepts: File I/O, string slicing, `while` loop, functions, conditionals, arithmetic**

The code below creates a CSV-style data file of student scores.  
Write a function `process_grades(filename)` that reads it and prints:
- Each student's name, score, and letter grade
- The class average
- The top-scoring student

**Grade scale:** A ≥ 90, B ≥ 80, C ≥ 70, D ≥ 60, F < 60

**File format (first line is a header):**
```
Name,Score
Alice,92
Bob,74
...
```


````python
# Creates the grades file
grades_data = """Name,Score
Alice,92
Bob,74
Charlie,88
Diana,61
Ethan,95
Fiona,55
George,83"""

with open("grades.csv", "w") as f:
    f.write(grades_data)

print("grades.csv created.")
````

````python
# ✅ Solution — Bonus

def letter_grade(score):
    """(int) -> str  Returns letter grade for a numeric score."""
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    elif score >= 60:
        return 'D'
    else:
        return 'F'


def process_grades(filename):
    """
    (str) -> None
    Reads a Name,Score CSV file and prints each student's grade,
    the class average, and the top scorer.
    """
    total_score  = 0
    student_count = 0
    top_name     = ''
    top_score    = -1
    
    with open(filename, 'r') as f:
        # Skip the header line
        header = f.readline()
        
        print(f"{'Name':<10} {'Score':>5}  {'Grade':>5}")
        print('-' * 24)
        
        for line in f:
            line = line.strip()
            if line == '':
                continue
            
            # Parse: find the comma separating name and score
            comma_idx = line.find(',')
            name  = line[:comma_idx]
            score = int(line[comma_idx + 1:])
            grade = letter_grade(score)
            
            print(f"{name:<10} {score:>5}  {grade:>5}")
            
            total_score   += score
            student_count += 1
            
            if score > top_score:
                top_score = score
                top_name  = name
    
    average = round(total_score / student_count, 2)
    
    print('-' * 24)
    print("Class average : " + str(average))
    print("Top student   : " + top_name + " (" + str(top_score) + ")")


process_grades("grades.csv")
````
