### String Comparisons
Strings are compared based on their **ASCII/Unicode values**:

- `==` checks if two strings are exactly the same.

**ASCII/Unicode values** are as follows:



Here are some examples:

# Section 3: Iteration using `while` loops



## When to use it?

- Use a `while` loop when you want a piece of code to be executed repeatedly as long as a particular condition is `True`.


## Syntax (general structure) of a `while` loop:

```python
while condition:
    body  # Must be indented!
```


- The condition is checked before each iteration.
- If the condition is False, the loop stops.
- Do not forget the colon : after the condition!

#### Example:
```python
x = 3
y = 1.5

while x < 4 and y < 4:
    print("Hello")
    y *= 2
```

How many times will this while loop execute?

````python
x = 3
y = 1.5

counter = 0  # Track how many times the loop runs
while x < 4 and y < 4:
    print("Hello")
    y *= 2
    counter += 1

print(f"The loop executed {counter} times.")
````

## Lazy Evaluation in `while` Loops


### What is Lazy Evaluation?
- Just like in `if` statements, when you use `and` or `or` in a `while` loop condition, the evaluation is **lazy**.
- This means **the second condition is only checked if necessary**.
- For example, in the condition `x < 4 and y < 4`, **Python only checks `y < 4` if `x < 4` is `True`**.




## Order of Conditions in `while` Loops

### Understanding Condition Evaluation Order

When using multiple conditions in a `while` loop with `and` or `or`, **Python evaluates them from left to right**. If Python determines the final outcome from the first condition alone, it does **not** evaluate the second condition (lazy evaluation).

### **Examples:**


#### **Case 1:**


````python

def some_function(x):
    print("function", x)
    return True

x = 12

while x > 10 and some_function(x):
    print("while", x)
    x = x - 1
````

#### **Case 2:**

````python
def some_function(x):
    print("function", x)
    return True

x = 12

while some_function(x) and x > 10:
    print("while", x)
    x = x - 1
````

Takeaways:

- Condition order affects execution.
- Expensive function calls should be placed after simpler conditions (if possible) to avoid unnecessary computations.
- Lazy evaluation prevents unnecessary function calls when using and or or.

## Notes on `while` Loops



### When to Use a `while` Loop?
- A `while` loop is useful when you **don't know how many iterations** will occur, but you **know when the iteration should stop**.

### Important Considerations:
- Make sure to **update the variables** used in the condition inside the loop body.
- If the condition **never becomes `False`**, the loop will result in an **infinite loop**.

## String Slicing

We can extract more than one character (or substring) using slicing. A slice is a substring from the start index up to **but not including** the end index.

#### Syntax:
- `String[start : end]`, e.g., `"Hello"[0:2]` returns `"He"`
- `String[start : end : step]`, e.g., `"Toronto"[0:5:2]` returns `"Trn"`

#### When either start or end are missing:
- `String[:i]` is same as `String[0:i]`
- `String[i:]` is same as `String[i:len(String)]`

Here are some examples:

````python
# Start index is missing
String = "Hello"
print(String[:2])
print(String[0:2])

# End index is missing
String = "Hello"
print(String[3:])
print(String[3:len(String)])

print(len(String))
````

#### Negative Steps

When using a negative step, slicing moves <b>backward</b>. This means that if you specify both `start` and `end`, the `start` index must be to the right of the `end` index.

````python
# Extract "cat" from String
String = "Spectacular"

print("Does this work? 1: ", String[4:7:-1]) 

# Below works! Start at position 6 ('c'), going backward 
# and stop at position 4 (= 3 + 1) ('t')
print("Does this work? 2: ", String[6:3:-1])
````

## String Immutability

Strings are <b>IMMUTABLE</b>: Slicing and indexing <b>DO NOT modify</b> the string they are acting on.

Try the following example:

````python
s = 'YOLO'
s[0] = 'P' # You CANNOT change Y to P!
````

### How to "modify" a string? $\rightarrow$ You CAN'T modify strings!

To update part of a string,

<b>Case 1: </b>You want to preserve the existing string value (a.k.a. the value to be "modified")
1. Create a new string using the "old" string
2. Assign the new string to a "new" variable

````python
string = "This is hard"

# Case 1
new_string = string[0:8] + "easy!"
print("new string      : ", new_string)

# We can still reference the original string value
print("original string : ", string)
````

1. Create a new string using the "old" string
2. Assign the new string to the "old" variable

## String Methods and Method Chaining

A <b>method</b> is a special type of function that is associated with a particular type of object (= a method is something that an object can do).

General syntax: `object_name.method_name(arguments)`

Here are some examples of methods for string type objects:

````python
name = 'Christine'

print('lower                : ', name.lower())
print('upper                : ', name.upper())
print('find the index of i  : ', name.find('i'))
print('rfind the index of i : ', name.rfind('i'))
print('replace i with y     : ', name.replace('i', 'y'))
print('name                 : ', name) # The original string remains same!
````

````python
# Methods can be chained
name = 'Christine'

print(name.lower().replace('ch', 'k'))
````
