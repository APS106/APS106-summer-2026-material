````python
# Linear transformation --> y = m*x + b
data = "10,45,12,999,105,111,124"
m = 3
b = 7

def linear_transformation(s, m, b):
    transformed_data = ""
    while s.find(',') != -1:
        
        idx = s.find(',')
        val = int(s[:idx])
        
        transformed_data += str(m*val + b) + ','
        
        s = s[idx+1:]
        
    transformed_data += str(m*int(s) + b)
    
    return transformed_data

print(linear_transformation(data, m, b))
````
