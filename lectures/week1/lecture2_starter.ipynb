### Terminology:

Let's pick an example you would know. `abs()` returns the absolute value of a number. 

In the example below `abs` is the **`function`** we are **`calling`** and `-20` is the **`argument`** that we're passing to the function. Same as before, the right side of the = gets evaluated first, and then the value gets assigned to the variable. In this case, the function **`returns`** `20`, so the variable x now stores the value 20.

````python
x = abs(-20)
print(x)
````

In the example below, you can see we're using two functions calls in the same line of code.

````python
y = abs(-20) + abs(-3)
print(y)
````

Let's make it a bit more complicated.

````python
y = -20
x = abs(y)
print(x)
````

Now the argument of the function call is not a constant (`-20`) but rather a variable (`y`). The rules for function calls are that each argument needs to be evaluated and then the function is called. 

1. `y` gets evaluated to its value (-20) 
2. -20 is passed to the function
3. The function is evaluated and returns 20
4. 20 is assignned to the variable `x`

How about this?

````python
y = -20
z = -5
x = abs(y + z)
print(x)
````

The thing in the parenthesis is an expression! And so it can be evaluated following the normal rules. That means that:
1. `y` is evalulated to -20
2. `z` is evaluated to -5
3. the `+` operator is evaluated with -20 and -5 to result in -25
4. -25 is passed to the `abs` function
5. 25 is returned from the `abs` function
6. 25 is assigned to `x`

Also, because,

1. **arguments of functions are expressions** 

2. **function calls are just expressions**

we can call functions with the output of other functions like this: (called a nested function call)

````python
x = pow(abs(-5), abs(-2))
print(x)
````

Remember: before a function is called, its arguments must be all evaluated. 

And so what happens above?

1. `abs()` is called with the value -5 and returns 5
2. `abs()` is called with the value -2 and returns 2
3. `pow()` is called with the values 5 and 2 and returns 25

So, `x = pow(int(5.2), abs(-2))` is the same as:

````python
x = pow(5, 2)
print(x)
````

## 6. Variable scope

A variable is only available from inside the region (indentation area) it is created; this is called the variable's scope

Local
- Whenever you define a variable within a function, its scope lies only within the function
- It is accessible from the point at which it is defined until the end of the function and exists for as long as the function is executing
- This means its value cannot be changed or even access from outside the function
- This is why we use `return` statements!

Global

- Whenever a variable is defined outside any function, it becomes a global variable and its scope is anywhere within the program
- This means that variables and functions defined outside of a function are accessible inside of a function
- If however your function has a local variable with the same variable name, it will use that value

In the code below, we have a variable `x` defined outside the function and a variable `x` in the function.

````python
def square(x):    
    print("Inside function. x = ", x)
    return x*x

x = 4
print("Outside function. x = ", x)
num_sq = square(8)
print("Outside function. x = ", x)
print(x, num_sq)
````

**The `x` variable inside the function and the `x` variable outside the function are completely independent!**

You should think variables named in the arguments of a function and those created inside the function as living **only** in side the function. The variable does not exist before the function is called. It is created when the function is called and it is destroyed with the function ends.

In the example below, we are not able to access temp outside of the function:

````python
def square(x):    
    temp = 5
    return x*x

num_sq = square(8)
print(temp)
````

## `return` v.s. `print`
    
`return` is the way to tell Python, "this function is now done, you can go back to the spot where this function was called" while also providing whatever result value the function made. 

If you want access to a value from a function, you must return it. If it is not included, it is the same as writing return `None`.

The function below prints the variable `output` but does not return it. 

````python
def square(x):
    output = x * x
    print(output)
    
square(2)
````

It may look like this function returns something but let's check.

````python
num = square(2)
print(num)
````

This means that if we don't add a `return` statement, Python will add one for us and return `None`.

````python
def square(x):
    output = x * x
    print(output)
    return None # This is always here, we just don't have to write it
````

If we rewrite the function and return the variable `output` instead of printing it, we get the following.

````python
def square(x):
    output = x * x
    return output
````

You can see the output looks slightly different. The integer `4` gets displayed but Jupyter shows a red `Out[ ]` display to the left of the output.

````python
square(2)
````

More importantly, we are able to capture the value that is returned by the function and assign it to a variable called `num`. This let's us use the output in some way elsewhere in our code. But, when we just print the output, we cannot capture that information and use it elsewhere.

````python
num = square(2)
print(num)
````

For example, let's say I want to use the square function below to square a number and then in the next line of code, I want to divide that number by 10.

````python
def square(x):
    output = x * x
    print(output)
````

````python
num = square(2)
````

````python
answer = num / 10 
````

I didn't include a return statement so Python returned `None` for me, which we cannot divide by 10.

Below is how to do this correctly.

````python
def square(x):
    output = x * x
    return output
````

````python
num = 4
````

````python
answer = num / 10 
print(answer)
````
