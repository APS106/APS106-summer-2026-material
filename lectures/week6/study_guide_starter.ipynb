  **Main Concepts:** Methods and updating.

# Breakout Session

Write a function named merge_lists that takes two dictionaries, dict1 and dict2, where keys are strings and values are lists of integers. The function should:

- Return a dictionary containing all keys from both dictionaries
- If a key appears in both dictionaries, the value in the new dictionary should be the concatenation of the two lists.
- Keys that appear in only one dictionary keep their original list.

```python
# Write code here
```

```python
print(merge_lists({"a": [1, 2], "b": [3]}, {"b": [4, 5], "c": [6]}))
# {"a": [1, 2], "b": [3, 4, 5], "c": [6]}

print(merge_lists({"x": [7]}, {"y": [8, 9], "x": [10]}))
# {"x": [7, 10], "y": [8, 9]}

print(merge_lists({}, {"a": [3, 4]}))
# {"a": [3, 4]}
```
